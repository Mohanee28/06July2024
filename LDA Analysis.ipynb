```python
x = df.drop(['Play'], axis = 1)
y = df['Play']
```

```python
sc = StandardScaler()
```

```python
print(x)
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
LDA = LinearDiscriminantAnalysis(n_components=1)
```

```python
X = LDA.fit_transform(x,y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state = 30)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lg = LogisticRegression()
```

```python
lg.fit(x_train, y_train)
```

```python
lg_prid = lg.predict(x_test)
```

```python
from sklearn.metrics import *
```

```python
accuracy_scor = accuracy_score(lg_prid, y_test)
```

```python
print(accuracy_scor)
```

```python
print('r2_score:', round(r2_score(lg_prid,y_test),4))
```

```python
print('MSE:', round(mean_squared_error(lg_prid, y_test), 4))
```

```python
print('MOE:', round(mean_absolute_error(lg_prid, y_test), 4))
```
